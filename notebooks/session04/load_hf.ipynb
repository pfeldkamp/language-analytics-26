````python
# How can we do this elegantly?
dfs = []

for key in ds.keys():
    print(key)
    tmp = ds[key].to_pandas()
    print("len of set:", len(tmp))
    dfs.append(tmp)

final = pd.concat(dfs)
print("final length:", len(final))
final.head()
````

````python
# let's make everything a function

def load_and_concat(dataset_name):
    # loading the dataset with the name
    ds = load_dataset(dataset_name)
    # see what sets are in there
    print("Keys:", ds.keys())
    # print("Len of sets:", [len(ds[key]) for key in ds.keys()])
    # collect all sets to pandas dfs in a list
    dfs = [ds[key].to_pandas() for key in ds.keys()]
    # concatenate
    final = pd.concat(dfs)
    print("Final len:", len(final))
    # return the result
    return final
    
````

````python

df = load_and_concat("cornell-movie-review-data/rotten_tomatoes")
df.head()
````
